## 1.  Linear Regression Algorithm and Gradient Descent Explain


* **Linear Regression** : Linear regression is a statistical method used to model the relationship between a       dependent variable and one or more independent variables by fitting a linear equation to the observed data. It's often used for making predictions or understanding the influence of independent variables on the dependent variable.
---

First recall that the equation of a line is:
$$ L(x) = mx + b $$
Where, $m$ and $b$ are the slope and the y-intercept respectively.  
---

However, in this project, it will need to change the formula to accept multiple $x$ variable. <br>
Overall, here is equation for linear regression:
$$ L(x) = \sum^{N}_{i=1} w_ix_i + b $$

$$L(x) = [w_1, w_2, w_3, \ldots, w_n] \cdot [x_1, x_2, x_3, \ldots, x_n] + b$$


**L(x)**: This represents the predicted or estimated value of the dependent variable (usually denoted as y) for a given set of independent variables x.<br>
<br>
$N$: It denotes the number of data points or observations in your dataset.<br>
<br>
**Σᵢ=₁ to N**: This is the summation symbol, indicating that you are summing up the following expression over all values of i from 1 to N. In other words, you're adding up all the terms for each data point in your dataset.<br>
<br>
**wᵢ**: These are the coefficients or weights associated with each independent variable. <br> 
<br>
**xᵢ**: These are the values of the independent variables for a specific data point i. Each xᵢ corresponds to one feature or input in your regression model.<br>
<br>
**b**: This is the intercept or bias term, representing the constant value that is added to the sum of the weighted independent variables. It shifts the regression line up or down along the dependent variable axis<br>

---

* **Gredient Descent** : Gradient Descent is an optimization algorithm used in machine learning and mathematical optimization. It's a method for finding the minimum of a function by iteratively adjusting parameters in the direction of steepest descent of the function's gradient. In the context of machine learning, it's commonly used to update model parameters during training to minimize a loss function and improve the model's performance.

The following is the notation that we will need to write down this important function. Since everything that we will be working with from here on out will be vectors of multiple features, we are going to be using gradient as the derivative we will be considering.
<br>
<br>
$$\nabla J(w_i,b) = \left \langle \frac{\partial J}{\partial m},\frac{\partial J}{\partial b} \right \rangle $$

So, let's compute this object:  
$$\nabla J(w_i,b) = \left \langle \frac{\partial}{\partial m} \left( \frac{1}{N} \sum^{N}_{i=1} (w_ix_i + b - y_i)^2 \right), \frac{\partial}{\partial b} \left( \frac{1}{N} \sum^{N}_{i=1} (w_ix_i + b - y_i)^2 \right) \right \rangle $$

$$= \left \langle \frac{2}{N} \sum^{N}_{i=1} (w_ix_i + b - y_i)x_i , \frac{2}{N} \sum^{N}_{i=1} (w_ix_i + b - y_i)  \right \rangle $$

1. $N$ is the number of datum we have **total**.
2. Each $x_i$ is one **feature** data value.
3. Each $y_i$ is the corresponding **target value**.
4. $w_i$ and $b$ are the **coefficients or weights associated with each independent variable** and **y-intercept**.
5. $w_ix_i + b$ is the **line**. (For the working $m$ and $b$)
6. $w_ix_i + b - y_i$ is the **error** between the line and the real data value.
   


### **Here is the code below**


```python
def MultiLinearRegression(train_x, Y, learning_Rate, epochs):

#     #get the column name for ploting
    x_name = train_x.columns
    Y_name = Y.name
    
    # create a list to store mse score
    mse_list = [] 
        
    # This is the N from above.
    N = len(Y)
    x = np.array(train_x)
    Y = np.array(Y)
    
     # Random weights for our slope and intercept.
    weights = np.zeros(x.shape[1]+1)
    
    y_intercept = [weights[1:]] # List to collect the y intercepts.
    slope = [weights[0]] # List to collect the slopes.
    
    ## BREAKDOWN OF EACH LINE BELOW.##
    for i in tqdm(range(epochs)):
        
    #Loop Over the equation to get all the prediction---------------
        pred = []
        for each_row in range(len(x)):
            predict =  np.dot(weights[1:],x[each_row]) + weights[0]
            pred.append(predict)
            
    #===============================================================
    
        error = pred - Y
        
        b = weights[0] - learning_Rate * (2 / N) * (error.sum()) #calculate the new y intercepts
        w = []
        
    # Loop for the new weights by applying the gradient formula
        for each_weight in range(len(weights[1:])):
            temp_w = weights[1:][each_weight] - learning_Rate * (2 / N) * ((x.T[each_weight] * error).sum())
            w.append(temp_w)
    #================================================================
    
        weights = [b]+w # combine b and w in a list
        
        y_intercept.append(b)
        slope.append(w)
        
        mse = MeanSquaredError(x,Y,weights[:1],weights[1:])   # calculate the MSE
        mse_list.append(mse)
    
    ###################################

    
# Ploting
    plt.figure(figsize=(9.5, 4)) # size of the graph
    
    # Create a plot shows MSE score
    ax = plt.subplot(121) # locate at the left side
    ax.plot(mse_list)     # plot the mse score history
    ax.set_xlabel('epoch')  
    ax.set_ylabel('Score')
    ax.set_title('MSE Score')
    
    #----------------------For single independent (2D) -------------------------------
    
    if x.shape[1] == 1:
        
        # matplotlib to show Linear Regression
        ax = plt.subplot(122)
        ax.scatter(x,Y)

        #plt.plot(x,y),y = weight_m * x + weight_b, color = red
        ax.plot(x, (weights[1] * x + weights[0]), c = "r")

        ax.set_xlabel(x_name[0])
        ax.set_ylabel(Y_name)
        print("The formula is y=",float(weights[1]),"x +",float(weights[0]))
        
    #--------------------For 2 independents (3D)---------------------------------------
        
    if x.shape[1] == 2: # to check if the data have two independents
        
        x_axis = np.array(train_x[x_name[0]])
        y_axis = np.array(train_x[x_name[1]])
        z_axis = Y

        # get the x and y surface data, generate number between the max and min of the og dataset
        x_surf, y_surf = np.meshgrid(np.linspace(x_axis.min(), x_axis.max(), len(data_2021))
                                     ,np.linspace(y_axis.min(), y_axis.max(), len(data_2021)))

        # get the Z surface by calculate with x surf and y surf
        line = []
        for row in range(len(x_surf.ravel())):
            each_dot = np.dot(weights[1:],[x_surf.ravel()[row],y_surf.ravel()[row]])+weights[:1]
            line.append(each_dot)

        ### Plotting
        ax = plt.subplot(122, projection='3d')
        ax.scatter(x_axis,y_axis,z_axis,c='red', marker='o', alpha=0.5)
        ax.plot_surface(x_surf,y_surf,np.array(line).reshape(x_surf.shape), color='b', alpha=0.3)
        ax.set_xlabel(x_name[0])
        ax.set_ylabel(x_name[1])
        ax.set_zlabel(Y_name)
        
        print("The formula is y= <",[weights],"> * <x_i> +",float(weights[0]))
        
    #-----------------------------------------------------------
    
    plt.tight_layout()
    plt.show()
    
    
    print("The mean square error is",mse_list[-1])         # print out the mean square error
    
    ###################################

    # Return the weights, the list of intercepts and the list of slopes.
    return [weights,y_intercept,slope]
```

* **Mean Square Error (MSE)**

**Mean Square Error (MSE)** is a commonly used metric to measure the average squared difference between the actual (observed) values and the predicted values in a regression problem. It quantifies the overall quality of a predictive model in terms of how well it fits the data. The MSE is defined as:

$$J(w_i,b) = \frac{1}{N} \sum^{N}_{i=1} (w_ix_i + b - y_i)^2  $$

**wᵢ**: These are the coefficients or weights associated with each independent variable xᵢ. They represent the importance or contribution of each independent variable to the prediction.<br>
<br>
**xᵢ**: These are the values of the independent variables for a specific data point i. Each xᵢ corresponds to one feature or input in your regression model.<br>
<br>
**b**: This is the intercept or bias term, representing the constant value that is added to the sum of the weighted independent variables. It shifts the regression line up or down along the dependent variable axis.<br>
<br>
**yᵢ**: These are the actual target values for each data point i. They represent the observed or true values of the dependent variable.

### **Here is the code below**

```python
def MeanSquaredError(x,y,b,w):
    mse_list = []                                       # a list to get all the result
    for i in range(len(x)):                             # start a for loop the same length as the x, which is 31
        Mse_formula = (np.dot(w,x[i]) + b - y[i])**2    # use the formula to get all the y - yi
        mse_list.append(Mse_formula)                    # put the resule in the list
    Mse = np.sum(mse_list)/len(x)                       # result / N(len(x))
    return Mse
```

In conclusion, linear regression is a powerful algorithm for predicting and visualizing data. The use of gradient descent as an optimization method greatly enhances the performance of linear regression models, preventing them from underfitting or overfitting the data. 

Our analysis of the World Happiness Score dataset underscores the importance of having a rich set of independent variables for the linear regression model. This dataset illustrates that the more independent variables included in the model, the more accurate the results tend to be. However, it's essential to consider the real-world implications of this approach.

In practice, incorporating more independent variables into a linear regression model increases the complexity of the research, as it requires additional data collection and resources. Moreover, there are trade-offs to consider. While a richer set of features may yield a more accurate model, it can also introduce noise and increase the risk of overfitting. Thus, the decision to include more independent variables in a linear regression model should be made judiciously, weighing the benefits against the associated costs and complexities.

In summary, linear regression, coupled with optimization techniques like gradient descent, provides valuable tools for data analysis and prediction. However, the selection of independent variables in the modeling process requires a careful balance between the desire for accuracy and the practical constraints of data collection and model complexity.
